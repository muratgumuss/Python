```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
# sentences
sentences = [
    'The glass of milk',
    'the glass of juice',
    'the cup of tea',
    'I am a good boy',
    'understand the meaning of words',
    'your videos are very good'
]
```

```python
# Define vocabulary size
vocab_size = 10000
```

```python
# One hot representation for every word in the sentence
one_hot_repr = [one_hot(sentence, vocab_size) for sentence in sentences]
one_hot_repr
```

```python
# word embedding representation

from tensorflow.keras.layers import Embedding
#from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
import numpy as np
```

```python
sentence_length = 8
embedded_doc = pad_sequences(one_hot_repr, maxlen=sentence_length, padding='pre')
embedded_doc

# feature representation
dim = 10
```

```python
model = Sequential()
model.add(Embedding(vocab_size, dim, input_length=sentence_length))
model.build(input_shape=(None, sentence_length))
model.compile('adam', 'mse')
```

```python
model.summary()
```

```python
model.predict(embedded_doc)
```
