````python
import os
from dotenv import load_dotenv
load_dotenv()

from langchain_groq import ChatGroq

groq_api_key = os.getenv("GROQ_API_KEY")
os.environ["HF_TOKEN"] = os.getenv("HF_TOKEN")

llm = ChatGroq(
    model="llama3-8b-8192",
    api_key=groq_api_key
)
llm
````

```markdown
2. LLM (Büyük Dil Modeli) Tanımlama
dotenv: .env dosyasındaki API anahtarlarını otomatik olarak yükler.
langchain_groq.ChatGroq: Groq firmasının LLM API'sine erişim sağlar. Burada "llama3-8b-8192" modeli kullanılıyor.
Amaç: Sorgulara cevap verecek büyük dil modeli hazırlanıyor.

```

````python
from langchain_huggingface import HuggingFaceEmbeddings
import os

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
````

```markdown
3. Embedding Modeli Oluşturma
langchain_huggingface.HuggingFaceEmbeddings: Hugging Face üzerinden cümleleri vektörlere (embedding) dönüştürür.
Amaç: Belgeleri ve sorguları sayısal vektörlere çevirerek benzerlik araması yapılmasını sağlar.
```

````python
# VectorStores
from langchain_chroma import Chroma

vectorstore = Chroma.from_documents(
    documents=documents,
    embedding=embeddings,
    persist_directory="chroma_db"
)

````

```markdown
4. Vektör Veritabanı (Vector Store) Oluşturma
langchain_chroma.Chroma: Chroma, vektör tabanlı arama için kullanılan bir veritabanıdır. Belgeler embedding ile vektörleştirilip burada saklanır.
Amaç: Belgeler vektör veritabanına kaydedilir, böylece hızlı benzerlik araması yapılabilir.
```

````python
vectorstore.similarity_search("LangChain nedir?")
````

```markdown
5. Benzerlik Araması
similarity_search: Sorgu cümlesini embedding'e çevirip, vektör veritabanında en benzer belgeleri bulur.
```

````python
# Asynchronous query
await vectorstore.asimilarity_search("LangChain nedir?")
````

````python
vectorstore.similarity_search_with_score("LangChain nedir?")
````

````python
# Retrievers
from typing import List
from langchain_core.documents import Document
from langchain_core.runnables import RunnableLambda

retriever = RunnableLambda(vectorstore.similarity_search).bind(k = 1)
retriever.batch(["langchain","python"]) 
    
````

```markdown
6. Retriever Oluşturma
langchain_core.runnables.RunnableLambda: Bir fonksiyonu (burada similarity_search) zincirleme işlemlerde kullanılabilir hale getirir.
bind(k=1): Her sorgu için 1 adet en benzer belge döndürmesini sağlar.
batch([...]): Birden fazla sorguyu topluca işler.
```

````python
retriever2 = vectorstore.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 1}
)
retriever2.batch(["langchain","python"])
````

```markdown
Alternatif retriever:
as_retriever: Chroma'nın kendi retriever arayüzü, benzer şekilde çalışır.


```

````python
# Rag Chain
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnablePassthrough

message = """"
Soru: {question}
Cevap: {answer}"""

prompt = ChatPromptTemplate.from_messages(
    [("human", message)]
)

rag_chain = {"answer": retriever, "question": RunnablePassthrough()} | prompt | llm

response = rag_chain.invoke("LangChain nedir?")

print(response)
````

```markdown
7. RAG (Retrieval-Augmented Generation) Chain Kurulumu
ChatPromptTemplate: Prompt şablonu oluşturur. {question} ve {answer} ile dinamik olarak doldurulur.
RunnablePassthrough: Zincirdeki girdiyi olduğu gibi geçirir (ör. kullanıcı sorusu).
rag_chain:
Önce "question" anahtarına kullanıcı girdisi, "answer" anahtarına ise retriever sonucu atanır.
Sonra prompt şablonu doldurulur.
Son olarak LLM'e gönderilir ve cevap alınır.
| operatörü ile zincirleme (pipeline) yapılır.

8. Sorgu ve Sonuç
invoke: Zinciri çalıştırır. Soru önce retriever'a gider, en alakalı belge alınır, prompt doldurulur ve LLM'den cevap alınır.
Amaç: Kullanıcıdan gelen soruya, belgelerden alınan bilgiyle desteklenmiş bir cevap üretmek.
```

```markdown
Özet Akış
Belgeler hazırlanır ve vektör veritabanına eklenir.
Sorgu embedding'e çevrilip en alakalı belge bulunur.
Soru ve cevap prompt şablonuna yerleştirilir.
LLM, bu prompt ile çalıştırılır ve son kullanıcıya cevap döner.
```

```markdown
RAG (Retrieval-Augmented Generation), temel olarak bir dil modelinin (LLM) dış bilgiye erişerek daha doğru ve güncel cevaplar üretmesini sağlar. Bu dış bilgi genellikle iki şekilde olabilir:

Kendi Belgeleriniz (Private/Local Data):
Elinizdeki dökümanlar, PDF'ler, veritabanları, şirket içi bilgiler gibi kaynaklardan alınan bilgiler. Kodunuzda yaptığınız gibi, bu belgelerden embedding ile vektör veritabanı oluşturup, sorgu sırasında en alakalı belgeyi bulup LLM'e veriyorsunuz.
Bu da bir RAG sistemidir ve çok yaygın kullanılır.

İnternetten Canlı Bilgi (Web Search):
Google, Bing, Wikipedia API gibi kaynaklardan gerçek zamanlı bilgi çekmek de mümkündür. Bu durumda retriever kısmı internetten arama yapar, bulduğu bilgiyi LLM'e aktarır.
Bu da RAG'ın bir başka uygulamasıdır.

Sonuç:
RAG, illa ki internetten bilgi almak zorunda değildir. Elinizdeki dokümanlardan da bilgi çekip LLM'e aktarırsanız, bu da bir RAG sistemidir.
Yani, RAG = (Retriever ile dış bilgiye eriş) + (LLM ile cevap üret).
Dış bilgi ister internetten, ister kendi belgelerinizden olabilir.

Senin kodunda, RAG sistemi kendi hazırladığın dokümanlar üzerinden çalışıyor.
```
