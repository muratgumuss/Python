```python
## Modeling bank account
## Create a class called BankAccount
## It should have a balance attribute

class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited {amount}. New balance is {self.balance}")
    
    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient funds")
        else:
            self.balance -= amount
            print(f"Withdrew {amount}. New balance is {self.balance}")
    
    def get_balance(self):
        print(f"Balance is {self.balance}")

account = BankAccount("Murat", 5000000)
account.deposit(5000)
account.withdraw(1000)
account.get_balance()
```

```python
# Inheritance
class Animal:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def eat(self):
        print("eat")

    def sleep(self):
        print("sleep")

class Dog(Animal):
    def __init__(self, name, age, breed):
        super().__init__(name, age)
        self.breed = breed

    def bark(self):
        print("bark bark bark")

dog = Dog("Rex", 2, "Bulldog")
print(dog.name, dog.age, dog.breed)
dog.eat()
dog.sleep()
dog.bark()
```

```python
class Animal:
    def __init__(self,name):
        self.name = name
    def speak(self):
        print("Sub class must implement this method")

class Pet:
    def __init__(self,owner):
        self.owner = owner
    def play(self):
        print("Sub class must implement this method")

## Derived class
class Dog(Animal,Pet):
    def __init__(self, name,owner):
        Animal.__init__(self,name)
        Pet.__init__(self,owner)
        ##super().__init__(name,owner)  

    def speak(self):
        return f"{self.name} says Woof!"
    def play(self):
        print("Fetch")

dog = Dog("Rex","Murat")
print(dog.speak())
print(dog.owner)
dog.play()
```

```python
## Polymporphism
class Animal:
    def __init__(self,name):
        self.name = name
    def speak(self):
        raise NotImplementedError("Sub class must implement this method")

class Dog(Animal):
    def speak(self):
        return f"{self.name} says Woof!"
    
class Cat(Animal):
    def speak(self):
        return f"{self.name} says Meow!"

dog = Dog("Rex")
cat = Cat("Whiskers")
print(dog.speak())
cat.speak()


```

```python
class Shape:
    def area(self):
        return "Area of the shape"

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        return 3.14 * self.radius * self.radius

## Function that demonstrates polymorphism
def calculate_area(shape):
    print (f"Area of the shape is {shape.area()}")

rectangle = Rectangle(10, 20)
circle = Circle(5)
calculate_area(rectangle)
calculate_area(circle)
```

```python
## Polymorphism with abstract base class
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass

class Car(Vehicle):
    def start_engine(self):
        print("Car engine started")

class Truck(Vehicle):
    def start_engine(self):
        print("Truck engine started")

car = Car()
truck = Truck()
car.start_engine()
truck.start_engine()

```

```python
class Car:
    def __init__(self, make, model, year, color):
        self._make = make  # Protected attribute
        self._model = model  # Protected attribute
        self._year = year  # Protected attribute
        self._color = color  # Protected attribute

    # Getter and Setter for make
    @property
    def make(self):
        return self._make

    @make.setter
    def make(self, make):
        self._make = make

    # Getter and Setter for model
    @property
    def model(self):
        return self._model

    @model.setter
    def model(self, model):
        self._model = model

    # Getter and Setter for year
    @property
    def year(self):
        return self._year

    @year.setter
    def year(self, year):
        self._year = year

    # Getter and Setter for color
    @property
    def color(self):
        return self._color

    @color.setter
    def color(self, color):
        self._color = color


# Kullanım
car = Car("Audi", "A4", 2019, "blue")
print(car.make)  # Çıktı: Audi
print(car.model)  # Çıktı: A4
print(car.year)  # Çıktı: 2019
print(car.color)  # Çıktı: blue

car.make = "Mercedes"
car.model = "C300"
car.year = 2022
car.color = "red"

print(car.make)  # Çıktı: Mercedes
print(car.model)  # Çıktı: C300
print(car.year)  # Çıktı: 2022
print(car.color)  # Çıktı: red
```

```python
## Access and modify private attributes
class Car:
    def __init__(self, make, model, year, color):
        self.make = make  # Public attribute
        self.model = model  # Public attribute
        self.year = year  # Public attribute
        self.__color = color  # Private attribute

    def get_color(self):
        return self.__color

    def set_color(self, color):
        self.__color = color
    
car = Car("Audi", "A4", 2019, "blue")
print(car.get_color()) # Output: blue
car.set_color("red")
print(car.get_color()) # Output: red
```

```python
## Abstract Base Class
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass

class Car(Vehicle):
    def start_engine(self):
        print("Car engine started")

class Truck(Vehicle):
    def start_engine(self):
        print("Truck engine started")

car = Car()
truck = Truck()
car.start_engine()
truck.start_engine()

    
```

```python
## Magic methods

class Car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    def __str__(self):
        return f"{self.make} {self.model} {self.year}"
    
    def __len__(self):
        return 10
    
    def __del__(self):
        print("Car object is deleted")

car = Car("Audi", "A4", 2019)
print(car.__str__())
print(car.__dir__())
print(car.__len__())
print(len(car))
del car
```
