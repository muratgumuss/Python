```python
## function definition
def my_function():
    print("Hello from a function")

## function call
my_function()

## function with arguments  
def my_function_with_args(username, greeting):
    print("Hello, %s , from a function, I wish you %s"%(username, greeting))

my_function_with_args("John Doe", "a great day")

## function with default arguments  
def my_function_with_default_args(username="John Doe", greeting="a great day"):
    print("Hello, %s , from a function, I wish you %s"%(username, greeting))

my_function_with_default_args()
my_function_with_default_args("Jane Doe", "a great day")

## function with return value
def sum(a, b):
    return a+b

print(sum(1, 2))
```

```python
## variable length arguments
def sum(*args):
    s = 0
    for i in args:
        s += i
    return s

print(sum(1, 2, 3, 4, 5))

## variable length keyword arguments
def print_kwargs(**kwargs):
    for key, value in kwargs.items():
        print("%s = %s"%(key, value))

print_kwargs(name="John Doe", age=30)

## lambda function
sum = lambda a, b: a+b
print(sum(1, 2))

## lambda function with map
numbers = [1, 2, 3, 4, 5]
squared = list(map(lambda x: x**2, numbers))
print(squared)

## lambda function with filter
even = list(filter(lambda x: x%2==0, numbers))  
print(even)

## lambda function with reduce
from functools import reduce
sum = reduce(lambda x, y: x+y, numbers)
print(sum)

## lambda function with sorted
sorted_numbers = sorted(numbers, key=lambda x: x%2)
print(sorted_numbers)

## lambda function with sorted
sorted_numbers = sorted(numbers, key=lambda x: x%2, reverse=True)
print(sorted_numbers)

## lambda function with sorted
sorted_numbers = sorted(numbers, key=lambda x: x%2, reverse=False)
print(sorted_numbers)

## lambda function with sorted
sorted_numbers = sorted(numbers, key=lambda x: x, reverse=True)
print(sorted_numbers)

## lambda function with sorted
sorted_numbers = sorted(numbers, key=lambda x: x, reverse=False)
print(sorted_numbers)

## lambda function with sorted
sorted_numbers = sorted(numbers, key=lambda x: x)
print(sorted_numbers)

## define function with list parameter
def print_list(l):
    for i in l:
        print(i)

print_list([1, 2, 3, 4, 5])
```
