```python
## porterstemmer
from nltk.stem import PorterStemmer

words = ["running", "ran", "easily", "fairness"]
# Create an instance of the PorterStemmer class
stemming = PorterStemmer()
for word in words:
    print(word, " : ", stemming.stem(word))

```

```python
from nltk.stem import RegexpStemmer

# Create an instance of the RegexpStemmer class with a custom pattern
regexp = RegexpStemmer('ing$|es$|ed$|able$', min=4)
words = ["running", "ran", "easily", "fairness","available"]
for word in words:
    print(word, " : ", regexp.stem(word))

```

```python
from nltk.stem import SnowballStemmer

## Create an instance of the SnowballStemmer class for English
snowball = SnowballStemmer("english")
words = ["running", "ran", "easily", "fairness"]
for word in words:
    print(word, " : ", snowball.stem(word))

```
