```python
import nltk

# Gerekli paketleri indir
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
nltk.download('maxent_ne_chunker')
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')
nltk.data.path.append(r'C:\Users\murat\AppData\Roaming\nltk_data')

# Named Entity Recognition (NER) using NLTK
sentences = (
    "My name is John Doe. I live in New York City. I work at OpenAI. "
    "I love programming in Python. "
    "I enjoy hiking in the mountains. "
    "I have a pet dog named Max."
)

# Cümleleri tokenize et
words = nltk.word_tokenize(sentences)

# Part-of-speech (POS) tagging
tag_elements = nltk.pos_tag(words)

# Named Entity Recognition (NER)
chunked = nltk.ne_chunk(tag_elements).draw()

# Sonucu yazdır
print(chunked)
```
