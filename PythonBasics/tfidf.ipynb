```python
# tf-idf example
from sklearn.feature_extraction.text import TfidfVectorizer

# Örnek metinler
corpus = [
    "I love programming",
    "Programming is fun",
    "I love learning new things"
]
```

```python
# TF-IDF modeli oluştur
tfidf_vectorizer = TfidfVectorizer()
X = tfidf_vectorizer.fit_transform(corpus)

# Kelime dağarcığını ve TF-IDF değerlerini görüntüle
print("Kelime Dağarcığı:", tfidf_vectorizer.get_feature_names_out())
print("Kelime Dağarcığı (vocabulary_):", tfidf_vectorizer.vocabulary_)

print("TF-IDF Matrisi:")
print(X.toarray())
```

```python
# TF-IDF modeli oluştur (unigram ve bigram için)
tfidf_vectorizer = TfidfVectorizer(ngram_range=(1, 2))  # 1: unigram, 2: bigram
X = tfidf_vectorizer.fit_transform(corpus)

# Kelime dağarcığını görüntüle
print("Kelime Dağarcığı (get_feature_names_out):", tfidf_vectorizer.get_feature_names_out())
print("Kelime Dağarcığı (vocabulary_):", tfidf_vectorizer.vocabulary_)

# TF-IDF değerlerini görüntüle
print("TF-IDF Matrisi:")
print(X.toarray())
```
