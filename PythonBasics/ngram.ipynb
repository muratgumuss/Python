```python
from sklearn.feature_extraction.text import CountVectorizer
# Örnek metinler
corpus = [
    "I love programming",
    "Programming is fun",
    "I love learning new things"
]
```

```python
# Bigram (2-gram) modeli
bigram_vectorizer = CountVectorizer(ngram_range=(2, 2))
X_bigram = bigram_vectorizer.fit_transform(corpus)
print("Bigram Kelime Dağarcığı:", bigram_vectorizer.get_feature_names_out())
print("Bigram Frekans Matrisi:")
print(X_bigram)
```

```python
# Trigram (3-gram) modeli
trigram_vectorizer = CountVectorizer(ngram_range=(3, 3))
X_trigram = trigram_vectorizer.fit_transform(corpus).toarray()
print("\nTrigram Kelime Dağarcığı:", trigram_vectorizer.get_feature_names_out())
print("Trigram Frekans Matrisi:")
print(X_trigram)
```
