```python
import nltk
import os
import shutil

# NLTK önbellek dizinini bul
cache_dir = os.path.join(os.path.expanduser('~'), '.nltk')
if os.path.exists(cache_dir):
    shutil.rmtree(cache_dir)
    print("NLTK önbellek dizini silindi.")

# nltk_data dizinini sil ve yeniden oluştur
nltk_data_path = r'C:\Users\murat\AppData\Roaming\nltk_data'
if os.path.exists(nltk_data_path):
    shutil.rmtree(nltk_data_path)
    print("Eski nltk_data dizini silindi.")

os.makedirs(nltk_data_path, exist_ok=True)
print("Yeni nltk_data dizini oluşturuldu.")

# Punkt paketini zorla indir (download_if_missing=True ile)
try:
    nltk.download('punkt', download_dir=nltk_data_path, download_if_missing=True)
    print("Punkt paketi başarıyla indirildi!")
except Exception as e:
    print("Hata:", e)

# İndirilen dosyaları kontrol et
if os.path.exists(nltk_data_path):
    print("Yeni indirilen dosyalar ve dizinler:")
    for item in os.listdir(nltk_data_path):
        full_path = os.path.join(nltk_data_path, item)
        print(f"{item} (dizin mi?: {os.path.isdir(full_path)})")
else:
    print("nltk_data dizini oluşturulmadı!")
```

```python
import nltk
from nltk.tokenize import sent_tokenize
nltk.download('punkt_tab')
# Veri yolunu ayarlayın
nltk.data.path = [r'C:\Users\murat\AppData\Roaming\nltk_data']

# Türkçe test
turkish_corpus = "Bu bir test cümlesidir. İşte başka bir cümle."
turkish_corpus = turkish_corpus.strip().lower()

try:
    turkish_sentences = sent_tokenize(turkish_corpus, language="turkish")
    print("Türkçe cümleler:", turkish_sentences)
except Exception as e:
    print("Türkçe için hata:", e)

# İngilizce test
english_corpus = "This is a test sentence. Here is another sentence."
english_corpus = english_corpus.strip().lower()

try:
    english_sentences = sent_tokenize(english_corpus, language="english")
    print("İngilizce cümleler:", english_sentences)
except Exception as e:
    print("İngilizce için hata:", e)
```

```python
corpus="""Hello Welcome,to Krish Naik's NLP Tutorials.
Please do watch the entire course! to become expert in NLP.
"""
##  Tokenization
## Sentence-->paragraphs
from nltk.tokenize import sent_tokenize
documents=sent_tokenize(corpus)
for sentence in documents:
    print(sentence)

## Tokenization 
## Paragraph-->words
## sentence--->words
from nltk.tokenize import word_tokenize

word_tokenize(corpus)
```
