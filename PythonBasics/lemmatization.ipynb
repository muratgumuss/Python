```python
import nltk
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
lemmatizer.lemmatize("running", pos="v")  # Specify the part of speech as a verb
lemmatizer.lemmatize("running", pos="n")  # Specify the part of speech as a noun
lemmatizer.lemmatize("running")  # Default is noun if not specified
lemmatizer.lemmatize("running", pos="a")  # Specify the part of speech as an adjective
lemmatizer.lemmatize("running", pos="r")  # Specify the part of speech as an adverb
lemmatizer.lemmatize("running", pos="s")  # Specify the part of speech as a satellite adjective
```

```python
from nltk.stem import WordNetLemmatizer
words = ["running", "ran", "runs", "run"]
for word in words:
    print(f"Lemmatized form of '{word}': {lemmatizer.lemmatize(word,pos="v")}")

```
