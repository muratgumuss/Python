```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# NLTK veri dizinini manuel olarak ayarla
nltk.data.path = ['C:\\Users\\murat\\AppData\\Roaming\\nltk_data']

# Gerekli modelleri indir
nltk.download('averaged_perceptron_tagger')
nltk.download('punkt')
nltk.download('stopwords')

# Örnek cümleler
sentences = ["This is a test sentence.", "NLTK is a powerful library for NLP."]

# POS etiketleme
for i in range(len(sentences)):
    sentences[i] = sentences[i].lower()
    words = word_tokenize(sentences[i])
    print(f"Tokenized words: {words}")  # Kelime listesini yazdır
    words = [word for word in words if word not in stopwords.words('english')]
    print(f"Filtered words: {words}")  # Stopwords çıkarıldıktan sonra listeyi yazdır
    pos_tags = nltk.pos_tag(words)
    print(f"POS tags: {pos_tags}")  # POS etiketlerini yazdır
    sentences[i] = pos_tags
```
