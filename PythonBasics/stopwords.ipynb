```python
## stop words
import nltk
nltk.download('stopwords') 
```

```python

from nltk.corpus import stopwords
stopwords.words('indonesian')
```

```python
from nltk.stem import SnowballStemmer
stemmer = SnowballStemmer("english")

paragraph = "The quick brown fox jumps over the lazy dog. The dog barked at the fox. The fox ran away.  " \
"The dog chased the fox. The fox climbed a tree. The dog couldn't reach the fox. "\
"The fox sat on a branch and laughed at the dog. The dog barked and jumped, but it couldn't reach the fox. " \

sentences = nltk.sent_tokenize(paragraph)
stemmer.stem(paragraph)

## apply stop words and stemming
for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    words = [stemmer.stem(word) for word in words if word not in stopwords.words('english')]
    sentences[i] = ' '.join(words)

sentences
```

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

paragraph = "The quick brown fox jumps over the lazy dog. The dog barked at the fox. The fox ran away.  " \
"The dog chased the fox. The fox climbed a tree. The dog couldn't reach the fox. "\
"The fox sat on a branch and laughed at the dog. The dog barked and jumped, but it couldn't reach the fox. " \

sentences = nltk.sent_tokenize(paragraph)
lemmatizer.lemmatize(paragraph)

## apply stop words and lemmatizer
for i in range(len(sentences)):
    sentences[i] = sentences[i].lower()
    words = nltk.word_tokenize(sentences[i])
    words = [lemmatizer.lemmatize(word) for word in words if word not in stopwords.words('english')]
    sentences[i] = ' '.join(words)

sentences
```
