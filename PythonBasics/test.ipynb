```python
# List 
my_list = [1,2,3,4,5]
print(my_list)
print(my_list[0])
print(my_list[1:3])
print(my_list[1:])
print(my_list[:3])
print(my_list[-1])
print(my_list[-2])
print(my_list[-3:-1])
print(my_list[0:5:2])
```

```python
mixed_list = [1, 2.5, "Hello", True]
print(mixed_list)
mixed_list.append(5)
print(mixed_list)
mixed_list.insert(2, "World")
print(mixed_list)
mixed_list.remove(2.5)
print(mixed_list)
mixed_list.pop()
print(mixed_list)
mixed_list.pop(1)
print(mixed_list)
```

```python
#iterating with index 
for index, value in enumerate(my_list):
    print(index, value)
```

```python
## list comprehension
list1 = [1,2,3,4,5]
list2 = [x**2 for x in list1]
pair = [(x,y) for x in list1 for y in list2]
print(pair)
```

```python
## list comprehension with function calls
words = ["hello", "world", "python", "programming"]
length = [len(word) for word in words]
print(length)
```

```python
## Creating Dictionary
my_dict = {"name": "John", "age": 20, "city": "New York"}
print(my_dict)
print(my_dict["name"])
empty_dict = {}
empty_dict["name"] = "John"
empty_dict["age"] = 20
empty_dict["city"] = "New York"
print(empty_dict)
dict1 = dict(name="John", age=20, city="New York")
print(dict1)
dict1["age"] = 25
print(dict1)
dict1["country"] = "USA"
print(dict1)
del dict1["city"]
print(dict1)
print(dict1.keys())
print(dict1.values())
print(dict1.items())
print("name" in dict1)
print("city" in dict1)
print(dict1.get("name"))
print(dict1.get("city","bulunmamaktadir"))
```

```python
## iterating over key value pairs
for key, value in dict1.items():
    print(key, value)
```

```python
## nested dictionary
my_dict = {"name": "John", "age": 20, "city": "New York", "courses": ["Math", "Science"]}
print(my_dict)
## adding a dictionary to a dictionary
my_dict["address"] = {"street": "123 Main St", "city": "New York"}
print(my_dict)
print(my_dict["address"]["city"])
```

```python
## Merge two dictionaries into one
dict1 = {"name": "John", "age": 20}
dict2 = {"city": "New York", "country": "USA"}
dict1.update(dict2)
print(dict1)
dict3 = {**dict1, **dict2}
print(dict3)    
```
