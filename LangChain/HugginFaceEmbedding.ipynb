```markdown
2. Hugging Face Token'ını Ayarlama
Amaç: Hugging Face API anahtarını ortam değişkeni olarak ayarlamak.

```

````python
from langchain_huggingface import HuggingFaceEmbeddings
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
````

```markdown
3. Hugging Face Embedding Modelini Oluşturma
Amaç: Hugging Face destekli bir embedding (vektörleştirme) modeli başlatmak.
Kütüphane: HuggingFaceEmbeddings — Metinleri vektörlere dönüştürür.
```

````python
text = "This is a test sentence for embedding."
embedding = embeddings.embed_query(text)
print("Embedding:", embedding)
print("Embedding length:", len(embedding))
````

```markdown

Amaç: Bir metni embedding vektörüne çevirmek ve vektörün boyutunu görmek.
```
