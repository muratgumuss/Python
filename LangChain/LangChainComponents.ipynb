````python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ['OPENAI_API_KEY'] = os.getenv('OPENAI_API_KEY')
# Langsmith tracking
os.environ['LANGCHAIN_API_KEY'] = os.getenv('LANGCHAIN_API_KEY')
os.environ['LANGCHAIN_TRACING_V2'] = "true"
os.environ['LANGCHAIN_PROJECT'] = os.getenv('LANGCHAIN_PROJECT')
````

```markdown
LANGCHAIN_API_KEY: LangSmith hesabınızdan aldığınız API anahtarıdır. Bu anahtarı ayarlayarak LangSmith ile bağlantı kurarsınız.
LANGCHAIN_TRACING_V2 = "true": Bu değişkeni "true" yaparak LangSmith ile uygulamanızın izlenmesini (tracing) aktif edersiniz.
LANGCHAIN_PROJECT: LangSmith üzerinde işlemlerin hangi proje altında izleneceğini belirtir.
Bu ortam değişkenleri ayarlandığında, LangChain kodunuz otomatik olarak LangSmith'e izleme (trace) ve log gönderir.
Yani, kodda doğrudan import langsmith gibi bir satır olmasa da, bu ortam değişkenleri sayesinde LangSmith entegrasyonu aktif olur.

Kısacası:
LangSmith'i kullandığınızı, yukarıdaki ortam değişkenlerinin ayarlanmasından ve "LANGCHAIN_TRACING_V2"'nin "true" olmasından anlıyoruz.
```

````python
from langchain_openai import ChatOpenAI
llm = ChatOpenAI(model = "gpt-4o")
print(llm)
````

````python
# Input and get response from the LLM
response = llm.invoke("What is generative AI?")
print(response.content)
````

````python
# Chatpromt template
from langchain.prompts import ChatPromptTemplate
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are an expert AI Engineer. Provide me answers based on the questions."),
    ("user", "{question}")
])
prompt
````

````python
chain = prompt | llm
# Input and get response from the LLM using the prompt
response = chain.invoke({"question": "Can you tell me about LangSmith?"})
print(response.content)
````

````python
# Stroutput parser
from langchain_core.output_parsers import StrOutputParser
output_parser = StrOutputParser()
chain = prompt | llm | output_parser

response = chain.invoke({"question": "Can you tell me about LangSmith?"})
print(response)
````

```markdown
StrOutputParser, LangChain zincirlerinde (chain) modelden dönen çıktıyı saf bir string (metin) olarak almak için kullanılır.

Normalde LLM'den gelen cevap bir nesne (örneğin AIMessage gibi) olabilir. StrOutputParser ise bu nesneden sadece düz metni (string) çıkarır ve zincirin çıktısını sadeleştirir.

Ne işe yarar?
Modelden dönen karmaşık nesneleri otomatik olarak string'e çevirir.
Zincir sonunda sadece düz metin (cevap) almak istediğinizde kullanılır.
Özellikle başka bir işleme veya ekrana yazdırmaya uygun sade bir çıktı sağlar.

Kısacası:
StrOutputParser, model çıktısını sadeleştirip doğrudan string olarak almanızı sağlar.
```
