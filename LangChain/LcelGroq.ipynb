````python
import os
from dotenv import load_dotenv
load_dotenv()

import openai
openai.api_key = os.getenv("OPENAI_API_KEY")
groq_api_key = os.getenv("GROQ_API_KEY")
groq_api_key
````

```markdown
1. Ortam Değişkenlerini Yükleme
dotenv: .env dosyasındaki API anahtarlarını ve ayarları otomatik olarak yükler.
Amaç: API anahtarlarını kodda açıkça yazmadan güvenli şekilde kullanmak.
2. API Anahtarlarını Alma
os.getenv: Ortamdan anahtarları çeker.
openai: OpenAI API’sini kullanmak için anahtar atanır.
groq_api_key: Groq API anahtarı alınır
```

````python
from langchain_openai import ChatOpenAI
from langchain_groq import ChatGroq
model = ChatGroq(model="gemma2-9b-It", api_key=groq_api_key)
model
````

```markdown
3. Model Tanımlama
langchain_groq: Groq LLM’lerini (büyük dil modelleri) LangChain ile kullanmayı sağlar.
model: Groq üzerinde çalışan bir dil modeli nesnesi oluşturur.
```

````python
from langchain_core.messages import HumanMessage,SystemMessage
messages = [
    SystemMessage(content="Translate the following question English to Turkish."),
    HumanMessage(content="Where is the capital of Turkey?"),
] 
result = model.invoke(messages)
````

```markdown
4. Mesaj Hazırlama ve Modeli Çağırma

langchain_core.messages: Modelle konuşmak için mesaj formatları sağlar.
SystemMessage: Modelin genel davranışını belirler.
HumanMessage: Kullanıcıdan gelen mesajı temsil eder.
model.invoke: Mesajları modele gönderir ve yanıt alır.
```

````python
from langchain_core.output_parsers import StrOutputParser
parser = StrOutputParser()
parser.invoke(result)
````

```markdown
5. Çıktıyı Metne Çevirme
StrOutputParser: Modelden gelen yanıtı sade bir metne çevirir.

```

````python
# Using LCEL - chain the components together
chain = model | parser
result = chain.invoke(messages)
result
````

```markdown
LCEL (LangChain Expression Language): Farklı bileşenleri (model, parser vb.) zincir gibi birbirine bağlamayı sağlar.
Amaç: Mesajı modele gönder, sonucu otomatik olarak metne çevir.
```

````python
# Prompt Template
from langchain_core.prompts import ChatPromptTemplate

generic_template = "Translate the following into {language}:"
prompt = ChatPromptTemplate.from_messages(
    [
        ("system", generic_template),
        ("user", "{text}"),
    ]
)
prompt_value = prompt.invoke({"language": "Turkish", "text": "Where is the capital of Turkey?"})
result = model.invoke(prompt_value.messages)
print(result)
````

```markdown
7. Prompt Template Kullanımı
ChatPromptTemplate: Dinamik olarak doldurulabilen mesaj şablonları oluşturur.
prompt.invoke: Şablondaki değişkenleri ({language}, {text}) verilen değerlerle doldurur.
model.invoke: Hazırlanan mesajları modele gönderir.
```

````python
# Chaining with lcel
prompt_value.to_messages()
chain = prompt | model | parser
result = chain.invoke({"language": "Turkish", "text": "Where is the capital of Turkey?"})
result
````

```markdown
8. Tüm Akışı Zincirleme
prompt | model | parser:
Prompt şablonunu doldur,
Mesajı modele gönder,
Sonucu metne çevir.
Tek adımda tüm işlemleri yapar.


Kütüphanelerin Görevleri
dotenv: Ortam değişkenlerini yükler.
os: Ortam değişkenlerine erişir.
langchain_groq: Groq LLM’lerini LangChain ile kullanır.
langchain_core.messages: Mesaj formatları sağlar.
langchain_core.prompts: Dinamik prompt şablonları oluşturur.
langchain_core.output_parsers: Model çıktısını işler.
LCEL (| operatörü): Bileşenleri zincirler, işlemleri otomatikleştirir.

Kodunuz, bir çeviri prompt’u hazırlıyor, Groq LLM’e gönderiyor ve sonucu sade metin olarak alıyor. LCEL ile tüm adımlar zincirlenip tek seferde çalıştırılabiliyor.


Groq, burada büyük dil modeli (LLM) hizmeti sağlayan bir altyapıdır.
Senin kodunda Groq şu işe yarıyor:

Dil modeli (ör: gemma2-9b-It) Groq sunucuları üzerinde çalışıyor.
Senin yazdığın prompt ve mesajlar, Groq API’si üzerinden bu modele gönderiliyor.
Model, verilen görevi (ör: çeviri) yerine getirip yanıtı döndürüyor.
Yani:

Groq, OpenAI gibi bir LLM servis sağlayıcısıdır.
Kodunda, langchain_groq ile Groq’un sunduğu dil modellerini kullanıyorsun.
Modelin çalışması, yanıt üretmesi Groq’un altyapısı sayesinde oluyor.
Kısaca:
Groq, senin prompt’unu işleyen ve cevap üreten yapay zeka modelini barındıran bulut servisidir.
```
